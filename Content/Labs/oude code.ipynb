```python
# Q added to the water as a measure of difference in degrees, mass and c_water
# Q = m*c*deltaT
Q = np.zeros(len(T_water_0))
for measn in range(len(T_water_0)):
    deltaT = T_water_max[measn] - T_water_0[measn]
    mass = M_water[measn]
    Q[measn] = c_water*mass*deltaT

# This is the amount of heat that the blokjes took up
print(f'Amount of added heat: {Q}')
# Q = m*c*deltaT

c = np.zeros(len(T_water_0))
for measn in range(len(T_water_0)):
    deltaT = T_blokje_0 - T_water_max[measn]
    mass = M_blokjes[measn]
    c[measn] = Q[measn]/(mass*deltaT)

print(f'Calculated c: {c}')

def STD(array):
    mean = np.mean(array)
    sum = 0
    for element in array:
        sum += (element - mean)**2
    return np.sqrt(sum/len(array))


def CalculateC(start_T_blokje, Q):
    C = np.zeros(len(Q))
    for measn in range(len(Q)):
        deltaT = start_T_blokje - T_water_max[measn]
        mass = M_blokjes[measn]
        C[measn] = Q[measn]/(mass*deltaT)
    return C

T_test = np.linspace(np.max(T_water_max), 200, 100)
std_bin = np.zeros(len(T_test))

for test in range(len(T_test)):
    result = CalculateC(T_test[test], Q)
    print(f'Result!: {result}')
    std_bin[test] = STD(result)

print(std_bin)
plt.plot(T_test, std_bin)
```
